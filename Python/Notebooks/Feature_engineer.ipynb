# Feature Engineering

## Objective

Create new business features from the master dataset for analysis, visualization and dashboard development.

```python
import pandas as pd
```

```python
master_df = pd.read_csv(
    "../outputs/cleaned_data/master_dataset.csv"
)
```

```python
# Datetime Conversion

date_columns = [

"order_purchase_timestamp",

"order_approved_at",

"order_delivered_carrier_date",

"order_delivered_customer_date",

"order_estimated_delivery_date"

]

for col in date_columns:

    master_df[col] = pd.to_datetime(master_df[col])
```

```python
# Time Features

master_df["order_year"] = master_df["order_purchase_timestamp"].dt.year

master_df["order_month"] = master_df["order_purchase_timestamp"].dt.month

master_df["order_month_name"] = master_df["order_purchase_timestamp"].dt.month_name()

master_df["order_day"] = master_df["order_purchase_timestamp"].dt.day

master_df["order_day_name"] = master_df["order_purchase_timestamp"].dt.day_name()

master_df["order_week"] = master_df["order_purchase_timestamp"].dt.isocalendar().week

master_df["quarter"] = master_df["order_purchase_timestamp"].dt.quarter
```

```python
# Weekend

master_df["is_weekend"] = master_df[
    "order_day_name"
].isin(
    ["Saturday","Sunday"]
)
```

```python
# Delivery Days

master_df["delivery_days"] = (

master_df["order_delivered_customer_date"]

-

master_df["order_purchase_timestamp"]

).dt.days
```

```python
# Approval Days

master_df["approval_days"] = (

master_df["order_approved_at"]

-

master_df["order_purchase_timestamp"]

).dt.days
```

```python
#Estimated Delay

master_df["delivery_delay"] = (

master_df["order_delivered_customer_date"]

-

master_df["order_estimated_delivery_date"]

).dt.days
```

```python
# Delivery Status

master_df["delivery_status"] = master_df["delivery_delay"].apply(

lambda x:

"Late"

if x>0

else "On Time"

)
```

```python
# High Value Orders

master_df["high_value_order"] = master_df["payment_value"] > 500
```

```python
# Shipping Percentage

master_df["shipping_percentage"] = (

master_df["freight_value"]

/

master_df["price"]

)*100
```

```python
# Installment Category

master_df["installment_type"] = master_df["payment_installments"].apply(

lambda x:

"Single"

if x==1

else "EMI"

)
```

```python
# Review Category

def review(score):

    if score>=4:

        return "Positive"

    elif score==3:

        return "Neutral"

    else:

        return "Negative"

master_df["review_category"] = master_df["review_score"].apply(review)
```

```python
#Price Category

master_df["price_category"] = pd.cut(

master_df["price"],

bins=[0,50,200,500,10000],

labels=[

"Low",

"Medium",

"High",

"Premium"

]

)
```

```python
# Save Dataset

master_df.to_csv(

"../outputs/cleaned_data/featured_dataset.csv",

index=False

)
```
