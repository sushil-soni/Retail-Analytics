```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use("ggplot")

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12,6)
```

```python
df = pd.read_csv(
"../outputs/cleaned_data/featured_dataset.csv")
```

```python
print(df.head())
print(df.shape)
print(df.describe())
print(df.info())
```

```python
df.select_dtypes(
include="number"
).columns
```

```python
df.select_dtypes(
include="object"
).columns
```

```python
corr = df.corr(
numeric_only=True
)

corr
```

```python
plt.figure(figsize=(16,10))

sns.heatmap(
corr,
annot=True,
fmt=".2f",
cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.heatmap(
df.isnull(),
cbar=False
)

plt.title("Missing Values")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
df["price"],
bins=50,
kde=True
)

plt.title("Price Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
df["delivery_days"],
bins=40,
kde=True
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
x="review_score",
data=df
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
x="payment_type",
data=df
)

plt.xticks(rotation=30)

plt.show()
```

```python
top = df[
"product_category_name_english"
].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
x=top.values,
y=top.index
)

plt.show()
```

```python
monthly = df.groupby(
"order_month_name"
)["payment_value"].sum()
```

```python
plt.figure(figsize=(12,6))

monthly.plot(
kind="line",
marker="o"
)

plt.show()
```

```python
top = df.groupby(
"customer_state"
)["payment_value"].sum()

top = top.sort_values(
ascending=False
).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
x=top.index,
y=top.values
)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
x="price",
y="freight_value",
data=df
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
x=df["price"]
)

plt.show()
```

Observation

• Most orders are delivered successfully.

• Credit Card is the most used payment method.

• Few product categories contribute the highest revenue.

• Delivery time is concentrated around the average range.

• Price distribution is right-skewed.
