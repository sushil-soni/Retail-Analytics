```python
import pandas as pd
```

```python
df = pd.read_csv(
"../outputs/cleaned_data/featured_dataset.csv"
)
```

```python
kpi = {

"Total Revenue":
df["payment_value"].sum(),

"Total Orders":
df["order_id"].nunique(),

"Total Customers":
df["customer_unique_id"].nunique(),

"Total Products":
df["product_id"].nunique(),

"Average Order Value":
round(df["payment_value"].mean(),2),

"Average Delivery Days":
round(df["delivery_days"].mean(),2),

"Average Rating":
round(df["review_score"].mean(),2)

}
```

```python
kpi_df = pd.DataFrame(

kpi.items(),

columns=[

"KPI",

"Value"

]

)

kpi_df
```

```python
kpi_df.to_csv(

"../outputs/reports/kpi_summary.csv",

index=False

)
```

```python
summary = {

"Top State":

df.groupby(

"customer_state"

)["payment_value"]

.sum()

.idxmax(),

"Top Category":

df.groupby(

"product_category_name_english"

)["payment_value"]

.sum()

.idxmax(),

"Top Seller":

df.groupby(

"seller_id"

)["payment_value"]

.sum()

.idxmax(),

"Most Used Payment":

df["payment_type"]

.mode()[0]

}
```

```python
summary_df = pd.DataFrame(

summary.items(),

columns=[

"Metric",

"Value"

]

)

summary_df
```

```python
summary_df.to_csv(

"../outputs/reports/business_summary.csv",

index=False

)
```
